```python
from ultralytics import YOLO

# 1. Load pretrained COCO model
model = YOLO("yolov8n.pt")

# 2. Train on your CarDD dataset
results = model.train(
    data="/content/cardd_colab.yaml",
    epochs=80,              # you can later increase to 80–100
    imgsz=640,
    batch=16,               # if OOM, reduce to 8
    device=0,               # use GPU
    workers=2,
    patience=10,
    project="/content/drive/MyDrive/cardd_runs",
    name="yolov8n_cardd",
    verbose=True,
)
```
